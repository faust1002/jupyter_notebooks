# Zadoff-Chu sequences (DFT of Zadoff-Chu)

### Introduction

Zadoff-Chu is defined as $$\Large x_{u}[n] = e ^ \frac{-j \pi u n (n + 1)}{P}, n \le 0 < P$$ where u represents uth root and P is a prime number.

DFT of Zadoff-Chu is defined as $$\Large X_{u}[k] = \sum_{n = 0}^{P - 1}x_{u}[n]e^\frac{-j2 \pi kn}{P}, 0 \le k < P$$

The DFT of the uth root (u = 0) ZC sequence x u[n] of prime length P is equal to a conjugated and time-shifted version of x u[n], up to a constant factor $$\Large X_{u}[k] = x_{u}^{*}[u^{-1} k]X_{u}[0]$$

where $u^{-1}$ is defined as multiplicative inverse of $u$ modulo $P$, such as $(u^{-1} u) \mod P = 1$. In addition to that, $u$ and $P$ must be relatively prime. Otherwise, $u^{-1}$ will not exist. In case of Zadoff-Chu sequences, since $P$ is always a prime number, it implies that it is always relatively prime with $u$.

The proof of this can be found <a href="https://www.researchgate.net/publication/224408244_Efficient_computation_of_DFT_of_Zadoff-Chu_sequences">here</a>.

Calculate Zadoff-Chu sequence and its DFT directly and them print DFT

```python
L = 839
u = np.random.randint(1, L - 1) # Based on 3GPP TS 38.211, Table 6.3.3.1-3,
                                  # actual root sequence index goes between 1 and 837
n = np.arange(0, L)
zc = np.exp(-1.0j * np.pi * u * n * (n + 1) / L)
ZC = np.fft.fft(zc)
```

Next, let us fix find the multiplicative inverse of $u$ modulo $P$ using Extended Euclidean Algorithm.

```python
def extended_euclidean_algorithm(u, p):
    a = np.max([u, p])
    b = np.min([u, p])
    t1 = 0
    t2 = 1
    while b != 0:
        q = a // b
        r = a % b
        t = t1 - t2 * q
        a, b, t1, t2 = b, r, t2, t
    if t1 < 0:
        t1 += p
    return t1
```

Finally, let us generate DFT of Zadoff-Chu sequence using the aforementioned equation and compare it with the DFT calculated directly.

```python
u_inverse = extended_euclidean_algorithm(u, L)
k = np.arange(0, L)
indices = (u_inverse * k) % L
ZC_prime = ZC[0] * np.conj(zc[indices])

cmp = np.any(np.isclose(ZC, ZC_prime))
if cmp:
    print("Zadoff-chu sequences matches")
else:
    print("Zadoff-chu sequences does not match :(")
```
